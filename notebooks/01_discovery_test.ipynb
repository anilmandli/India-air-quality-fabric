```python
import requests
from google.colab import userdata

api_key = userdata.get('OPENAQ_KEY')
headers = {"X-API-Key": api_key}

# 1. Verify connection and get Country ID for India
res_countries = requests.get("https://api.openaq.org/v3/countries", headers=headers)
print("Countries Status:", res_countries.status_code)

if res_countries.status_code == 200:
    countries = res_countries.json().get('results', [])
    india = next((c for c in countries if c.get('name') == 'India'), None)
    print("India Details:", india)

# 2. Get Locations in India
res_locations = requests.get("https://api.openaq.org/v3/locations?iso=IN&limit=100", headers=headers)
print("Locations Status:", res_locations.status_code)

if res_locations.status_code == 200:
    locations = res_locations.json().get('results', [])
    print(f"Retrieved {len(locations)} locations in India.")

    # Print the first 5 locations (Name, Location ID, Sensors)
    for loc in locations[:5]:
        print(f"Location: {loc.get('name')} | ID: {loc.get('id')}")
        for sensor in loc.get('sensors', []):
            print(f"   -> Sensor ID: {sensor.get('id')} | Parameter: {sensor.get('parameter', {}).get('name')}")
```

```python
# Test pulling actual measurements for a sensor (e.g., Sensor ID 30 for IGI Airport PM2.5)
sensor_id = 30
res_measurements = requests.get(
    f"https://api.openaq.org/v3/sensors/{sensor_id}/measurements?limit=10",
    headers=headers
)

print("Measurements Status:", res_measurements.status_code)
if res_measurements.status_code == 200:
    measurements = res_measurements.json().get('results', [])
    print("Sample Measurement Data:", measurements[0] if measurements else "No data")
```
